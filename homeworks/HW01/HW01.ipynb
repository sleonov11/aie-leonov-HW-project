```python
#dispersion try

arr = np.random.randint(50,1000, 50)

def dispersion(arr):
    mean = np.mean(arr)
    n = np.size(arr)
    sum_ = 0
    for i in range(n):
        x = arr[i] - mean
        sum_ += x**2
    sigma = math.sqrt(sum_/n)
    return sigma
print(dispersion(arr),'\n',arr)


np.std(arr)
```

```python
mask = arr > 150
res = arr[mask]
print(res, '\n', 'количество эл-ов удовл усл', np.size(res))
```

```python
#mse
n = 50
def mse(n):
    y_true = np.random.randint(50,1000, 50)
    y_pred = np.random.randint(5,993, 50)
    sum = 0
    #нужно нормировать перед вычислением
    norm_y_true = (y_true - np.min(y_true)) / (np.max(y_true) - np.min(y_true))
    norm_y_pred = (y_pred - np.min(y_pred)) / (np.max(y_pred) - np.min(y_pred))
    for i in range(n):
        a = (norm_y_true[i] - norm_y_pred[i])**2
        sum += a
    res = sum / n
    return res
print('псевдо мсе:',mse(n), '\n', 'rmse:', math.sqrt(mse(n)))
def mse_without_loop(n):
    y_true = np.random.randint(50,1000, 50)
    y_pred = np.random.randint(5,993, 50)
    norm_y_true = (y_true - np.min(y_true)) / (np.max(y_true) - np.min(y_true))
    norm_y_pred = (y_pred - np.min(y_pred)) / (np.max(y_pred) - np.min(y_pred))
    return np.mean((norm_y_true - norm_y_pred)**2)
print(mse_without_loop(n))

#отличаются из- за разщных массивов,
#массивы генерируются локально в функциях
```
